```python
#=====================================================================================
# Calcule e exiba receitas de produtos usando funções, com resultados apresentados 
# de forma clara e formatada.
#=====================================================================================

#=================================================================
# Lista de produtos, seus preços e quantidades vendidas
#=================================================================
products = ["Pães", "Maçãs", "Laranjas", "Bananas", "Melãos", "Kiwi", "Caqui", "Mixiricas", "Cerejas (100 gr)"]
prices = [0.50, 1.20, 2.50, 2.00, 30.00, 4.00, 6.00, 1.00, 15.00]  # preço por item
quantities_sold = [150, 200, 100, 50, 25, 15, 120, 35, 10]  # Qtde de itens vendidos

print(" ")
print("=" * 41)
print("Res_001 - Relatório das frutas em estoque")
print("=" * 41)
print(" ")


def calculate_revenue(prices, quantities_sold):
    """Calcula a receita total de cada produto e o total geral"""
    
    revenue = []
    total_vendas = 0
    
    for i in range(len(prices)):
        valor = prices[i] * quantities_sold[i]
        revenue.append(valor)
        total_vendas += valor
    return revenue, total_vendas

def formatted_output(revenues):
    """Exibe relatório formatado"""
    for product, qtd, rev in sorted(revenues, key=lambda x: x[0]):  # ordena por produto
        print("-" * 55)
        print(f"O estoque de {qtd} {product} gerou uma receita de R${rev:.2f}")


# Chamadas principais
revenue, tot_vendas = calculate_revenue(prices, quantities_sold)
# Juntando produto + quantidade + receita
revenue_per_product = list(zip(products, quantities_sold, revenue))
formatted_output(revenue_per_product)

print("-" * 55)
print(f"Total das vendas R${tot_vendas:.2f}")
```
