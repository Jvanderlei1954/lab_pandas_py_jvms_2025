```python
#=============================================================================
# Processando: Cadastro de pessoas
#=============================================================================

import csv
import pandas as pd
import matplotlib.pyplot as plt  

# Nome do arquivo de entrada e saída
arquivo_entrada = 'general_data.csv'
arquivo_saida = 'ds_general.csv'
arquivo_calcula = 'ds_calcula.csv'

# Lendo o CSV com pandas
df = pd.read_csv(arquivo_entrada, encoding='utf-8')

# RENOMEAR COLUNAS DO DATASET general_data.csv sobre o dataframe df
df.rename(columns={
    'Age': 'Idade',
    'Attrition': 'Atrito',
    'BusinessTravel': 'Viagem',
    'Department': 'Depto',
    'DistanceFromHome': 'Distancia',
    'Education': 'Cursos',
    'EducationField': 'Area',
    'EmployeeCount': 'Qt_Func',
    'EmployeeID': 'ID',
    'Gender': 'Genero',
    'JobLevel': 'Nivel',
    'JobRole': 'Fucao',
    'MaritalStatus': 'Est_Civil',
    'MonthlyIncome': 'RE_Mensal',
    'NumCompaniesWorked': 'Empresas',
    'Over18': 'Adulto',
    'PercentSalaryHike': 'Aumentos',
    'StandardHours': 'TT_Horas',
    'StockOptionLevel': 'Op_Cargos',
    'TotalWorkingYears': 'Carreira',
    'TrainingTimesLastYear': 'Qt_Trein',
    'YearsAtCompany': 'Temp_Casa',
    'YearsSinceLastPromotion': 'Qt_Promo',
    'YearsWithCurrManager': 'Temp_Ger'   
}, inplace=True)

# Função para alterar conteúdo de uma coluna
def alterar_valor(df, coluna, valor_antigo, novo_valor):
    """
    Altera todos os valores iguais a `valor_antigo` na `coluna` informada para `novo_valor`.
    """
    df[coluna] = df[coluna].replace(valor_antigo, novo_valor)
    return df

# Exemplo de uso: Alterar 'Research & Development' para 'P&D' na coluna 'Depto'
df = alterar_valor(df, 'Atrito', 'Yes', 'Y')
df = alterar_valor(df, 'Atrito', 'No', 'N')

df = alterar_valor(df, 'Area', 'Life Sciences', 'Ciencias')
df = alterar_valor(df, 'Area', 'Medical', 'Medicina')
df = alterar_valor(df, 'Area', 'Other', 'Outros')
df = alterar_valor(df, 'Area', 'Technical Degree', 'Nivel_Tecnico')

df = alterar_valor(df, 'Viagem', 'Travel_Frequently', 'Sempre')
df = alterar_valor(df, 'Viagem', 'Travel_Rarely', 'Rara')
df = alterar_valor(df, 'Viagem', 'Non-Travel', 'Local')

df = alterar_valor(df, 'Genero', 'Female', 'Feminino.')
df = alterar_valor(df, 'Genero', 'Male', 'Masculino')

df = alterar_valor(df, 'Depto', 'Research & Development', 'P&D')
df = alterar_valor(df, 'Depto', 'Sales', 'Vendas')

print("Primeiras linhas do DataFrame:")
print(df.head(25))

# Convertendo o DataFrame em lista de listas, com strings em maiúsculas
dados_processados = [ [str(campo).upper() for campo in df.columns] ]  # cabeçalho
for _, linha in df.iterrows():
    linha_maiuscula = [str(campo).upper() for campo in linha]
    dados_processados.append(linha_maiuscula)
    
    
# ========================================================================
# CRIANDO ARQUIVO COM O CÁLCULO DA RENDA ANUAL
# ========================================================================
         
# Criar novo arquivo adicionando coluna para salário anual
df['RE_Anual'] = df['RE_Mensal'] * 12

# Criar coluna de faixa salarial com base no salário mensal
df['Faixa'] = df['RE_Mensal'].apply(lambda x: 'Alto' if x > 5000 else 'Médio' if x > 4000 else 'Baixo')

# Salvar o novo DataFrame no arquivo de saída
df.to_csv(arquivo_calcula, index=False, encoding='utf-8')

# Escrevendo no arquivo CSV de saída
with open(arquivo_saida, mode='w', newline='', encoding='utf-8') as saida:
    escritor = csv.writer(saida)
    escritor.writerows(dados_processados)

# ******************************************************
# Fazendo a impressão da relação dos nomes das colunas
# ******************************************************
print("=" * 60)
print("========= Colunas encontradas no dataset: ==================")
print("=" * 60)
for nome in df.columns:
    print(f"- {nome}")

print("*" * 73)    
print("Processamento concluído! Arquivo de saída criado como 'ds_general.csv'.")
print("*" * 73)
```
