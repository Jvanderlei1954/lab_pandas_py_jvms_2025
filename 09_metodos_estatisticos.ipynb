```python
#======================================================================================================
# Utilizando os métodos python para encontrar no dataframe os valores máximos, mínimos e o valor médio 
# no dataframe cars_data, combinados com média x mediana.
# =====================================================================================================
```

```python
# Importando biblioteca e trabalhando com os métodos estatísticos do python
#==============================================================================
import pandas as pd 

cars_data = {'model': ['audi A1', 'audi A6', 'audi A4', 'audi A3','audi A1'],
          'year': [2017, 2016, 2017, 2019, 2016],
          'fueltype': ['petrol', 'diesel', 'diesel', 'petrol', 'petrol'],
          'price': [12500, 16500, 16800, 17300, 16500]}

audi_cars = pd.DataFrame(cars_data)

# Write your code below
highest_price = audi_cars['price'].max()
lowest_price = audi_cars['price'].min()
average_price = audi_cars['price'].mean()

# Diferença entre faixas e médias com relação a variação dos preços
dif_maior_preco_e_media = (highest_price - average_price)
dif_menor_preco_e_media = (lowest_price - average_price) 
dif_maior_e_menor_preco = (highest_price - lowest_price)

# Percentuais obtidos através das variações dos preços
perc_maior_media = (dif_maior_preco_e_media / highest_price)
perc_media_menor = (dif_menor_preco_e_media / average_price )
perc_maior_menor = ((dif_maior_e_menor_preco / highest_price))


# Apresentação dos resultados
print(" ")
print(f' Sumarização: Maior Preço: {highest_price:.2f} | Menor preço: {lowest_price:.2f} | Preço Médio: {average_price:.2f}')
print("=" * 90)


print("")
print(" Apresentação detalhada dos resultados calculados: Média versos preços praticados")
print("=" * 90)
print(f'Maior Preço: {highest_price:.2f} | Dif. Maior preço/média: {dif_maior_preco_e_media:.2f} - Var: {perc_maior_media * 100:.2f}% (acima da média)')
print("=" * 90)

print(f'Menor Preço: {lowest_price:.2f} | Dif. Média/menor preço: {dif_menor_preco_e_media:.2f} - Var: {perc_media_menor * 100:.2f}% (abaixo da média)')
print("=" * 90)

print(f'Preço Médio: {average_price:.2f} | Dif. Maior/menor preço: {dif_maior_e_menor_preco:.2f} Var: {perc_maior_menor * 100:.2f}% (acima da média)') 
print("=" * 90)
#print('')


# Calcular a mediana apenas da coluna "price"
mediana = audi_cars['price'].median()

print(f"Preço mediano = {mediana:.2f} | Preço mediano x preço médio: {(mediana - average_price):.2f} - (Mediana maior que média)")
print("=" * 90)
print('')
print("=" * 35)
print('Visualização do arquivo audi_cars')
print('='* 35)
print(audi_cars.head(10))
print('=' * 35)

```
