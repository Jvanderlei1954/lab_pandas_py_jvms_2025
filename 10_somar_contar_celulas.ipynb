```python
# Práticas com contagem e somatórios de celulas dataframe pytho -> count() e sum()
```

```python
import pandas as pd 
import numpy as np

cars_data = {'model': [None, 'audi A6', 'audi A4', None,'audi A1'],
             'year': [2017, 2016, 2017, None, 2016],
             'fueltype': [None, 'diesel', 'diesel', 'petrol', 'petrol'],
             'price': [12500, 16500, 16800, 17300, 16500]}

audi_cars = pd.DataFrame(cars_data)

print('Visualização do arquivo audi_cars')
print('='* 35)
print(audi_cars.head(10))
print('=' * 35)

# Write your code below
number_of_cells = audi_cars.count()
total_price = audi_cars['price'].sum()
null_count = audi_cars.isna().sum()

# Testing the result
print('')
print(f'Registros com valores nulos:')
print('*' * 35)
print(null_count)
print('*' * 35)
print(" ")

print('*' * 35)
print('Registros válidos:')
print(number_of_cells)
print('*' * 35)

print(f'Somatória dos preços: {total_price}')
```
