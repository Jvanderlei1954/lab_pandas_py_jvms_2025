```python
import pandas as pd

# ============================================================
# 1. DADOS DA ACADEMIA
# ============================================================
planos = {
    "Mensal": {"preco": 423, "alunos": 80},      # R$ 423 por mês
    "Trimestral": {"preco": 1200, "alunos": 40},  # R$ 1200 a cada 3 meses
    "Anual": {"preco": 4308, "alunos": 20}       # R$ 4308 por ano
}

# Custos fixos mensais
custos_fixos = {
    "Aluguel": 5000,
    "Associados": 12000,
    "Energia": 350,
    "Água": 180,
    "Manutenção": 400,
    "Marketing": 500
    
}

# ============================================================
# 2. CÁLCULO DA RECEITA
# ============================================================
# Receita mensal estimada
receita_dados = []
for plano, dados in planos.items():
    if plano == "Trimestral":
        receita_mensal = (dados["preco"] / 3) * dados["alunos"]  # distribui por mês
    elif plano == "Anual":
        receita_mensal = (dados["preco"] / 12) * dados["alunos"] # distribui por mês
    else:
        receita_mensal = dados["preco"] * dados["alunos"]
    
    receita_dados.append([plano, dados["alunos"], dados["preco"], receita_mensal])

df_receita = pd.DataFrame(receita_dados, columns=["Plano", "Qtde Alunos", "Preço (R$)", "Receita Mensal (R$)"])

# Receita total mensal
receita_total = df_receita["Receita Mensal (R$)"].sum()

#print(f"Total dos tributos = {tot_tributos:.2f} ")
# ============================================================
# 3. CÁLCULO DOS CUSTOS
# ============================================================
df_custos = pd.DataFrame(list(custos_fixos.items()), columns=["Custo", "Valor (R$)"])
custo_total = df_custos["Valor (R$)"].sum()
tot_tributos = receita_total * 0.06

# ============================================================
# 4. RESULTADO FINAL
# ============================================================

#=============================================================
# 5. Inclusão do Cálculo do Pro-Labore no Resultado Final
#=============================================================
val_lab = (receita_total - custo_total) * 0.4 # Pró-Labore de 40%
desc_inss = (val_lab * 11/100)
desc_irpf = (val_lab * 14/100)
val_lab = (val_lab - desc_inss - desc_irpf)

part_lab1 = val_lab * 0.5
part_lab2 = val_lab * 0.4
lucro_liquido = (receita_total - custo_total) - (part_lab1 + part_lab2 + tot_tributos)
tot_prolab = (part_lab1 + part_lab2)

perc_liq = (lucro_liquido/receita_total) * 100
ind_desp_final = (100 - perc_liq)/100
desp_final = receita_total * ind_desp_final

#================================================================
# Imprimindo os resultados
#================================================================
print("="*56)
print(" RELATÓRIO FINANCEIRO - ACADEMIA DE PILATES ")
print("="*56)
print("\n--- Receita por Plano ---")
#*****************************************************************
print(df_receita.to_string(index=False))
print("=" * 56)
print("\nReceita Total Mensal: R$ {:,.2f}".format(receita_total))
print("\nDespesa Total Mensal: R$ {:,.2f}".format(desp_final))
print("\nPerc. Despesa Total: R$ {:,.2f}".format(ind_desp_final*100))
print("=" * 56)

#*****************************************************************
print("\n--- Custos Fixos ---")
print(df_custos.to_string(index=False))
#*****************************************************************
print(f"Tributos          {tot_tributos:.0f}")
print(f"Pró_Lab_Socio1    {part_lab1:.0f}")
print(f"Pró_Lab_Socio2    {part_lab2:.0f}")
print("=" * 56)
print("\nSomatório dos Pró-labores: R$ {:,.2f}".format(tot_prolab))
print("\nCustos Fixos sem Tributos e Pró-labore: R$ {:,.2f}".format(custo_total))
print("=" * 56)
print("\nCusto Fixo Mensal: R$ {:,.2f}".format(custo_total + part_lab1 + part_lab2 + tot_tributos))
print("=" * 56)
print("\n>>> Lucro Líquido Mensal: R$ {:,.2f}".format(lucro_liquido))
print(f">>> Lucro Líquido: {perc_liq:.2f}% da Receita Total Mensal")
print("=" * 56)
```
