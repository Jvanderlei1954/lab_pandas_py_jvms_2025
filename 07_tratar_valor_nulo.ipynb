```python
############################
# Preenchendo valores nulos

import pandas as pd

# Carregar o CSV
wine_data = pd.read_csv(
    'https://codefinity-content-media.s3.eu-west-1.amazonaws.com/a43d24b6-df61-4e11-9c90-5b36552b3437/wine_with_nan.csv'
)

# Preencher valores nulos com "no"
wine_data = wine_data.fillna("no")

# Exibir informações do DataFrame
if (wine_data['quality'] != 'no').any():
    print("A coluna 'quality' contém 'no'")
    print(wine_data)
else:
    print("Nenhum 'no' encontrado na coluna 'quality'")
    print(wine_data.info())


```
