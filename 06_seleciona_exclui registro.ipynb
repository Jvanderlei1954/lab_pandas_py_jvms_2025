```python
# Usando o comando dropna() Selecione, exclua e imprimima os registros com valores não nulos
```

```python
import pandas as pd

# Carregar dataset (nome do arquivo corrigido)
wine_data = pd.read_csv('https://codefinity-content-media.s3.eu-west-1.amazonaws.com/a43d24b6-df61-4e11-9c90-5b36552b3437/wine_with_nan.csv')

# Comprimento antes de remover nulos
length_before = len(wine_data)
print("Total de linhas antes da remoção:", length_before)

# Remover linhas com valores nulos
wine_data = wine_data.dropna()

# Comprimento após remover nulos
length_after = len(wine_data)
print("Total de linhas depois da remoção:", length_after)

# Imprimir número de linhas removidas
print("Total de linhas removidas:", length_before - length_after)
```
